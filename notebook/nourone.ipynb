```python
import pandas as pd
import numpy as np
import cv2
from pathlib import Path
import zipfile

import tensorflow as tf
# from google.colab import drive
from sklearn.model_selection import train_test_split
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
import time
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
import gdown
import os
from IPython.display import FileLink , display
import mlflow
from mlflow import MlflowClient
```

```python
VALID_EXTANTION = ['.jpeg','.jpg','png','bmp']

def load_data(data_dir,image_size=(224,224)):

  data_dir = Path(data_dir)

  if data_dir.is_file() and data_dir.suffix.lower() == ".zip":
    extract_dir = data_dir.with_suffix("")
    with zipfile.ZipFile(data_dir,"r") as zip_ref:
      zip_ref.extractall(extract_dir)

    data_dir = extract_dir

    data_dir = data_dir / "Data"

  if not data_dir.is_dir():
    raise FileNotFoundError(
        f"dossier introvable : {data_dir}"
    )

  images = []
  labels = []
  class_names = sorted(
    p.name for p in data_dir.iterdir() if p.is_dir()
    )

  if not class_names:
      raise ValueError(
          "Aucun class trouvé"
      )

  encoded = LabelEncoder()
  encoded.fit(class_names)

  for class_name in class_names:
      class_dir = data_dir / class_name

      for path in class_dir.rglob("*"):

        if not path.is_file():
          continue

        if path.suffix.lower() not in VALID_EXTANTION :
          continue

        try:
            image = cv2.imread(str(path))

            image = cv2.cvtColor(image , cv2.COLOR_BGR2RGB)

            image = cv2.resize(image , image_size)

            images.append(image)
            labels.append(class_name)

        except Exception as e:

            print(f"image ignor {path} : {e}")

  x = images
  y = encoded.transform(labels)

  print("number des images valid : ",len(x))
  print("classes : ",class_names)
  # print("dimension : ",x.shape)

  return x , y , class_names
```

```python
fill_id = "1a4w4HfwvyzmdIF56TmWL5-z8of0ghE3m"
Zip_Path = "data.zip"

url = f"https://drive.google.com/uc?id={fill_id}"

gdown.download(url,Zip_Path,quiet=False)
```

```python
x , y , class_names = load_data(Zip_Path)
```

```python
x_train , x_val_test , y_train ,y_val_test = train_test_split(
    x,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y

)

x_val , x_test , y_val , y_test = train_test_split(
    x_val_test,
    y_val_test,
    test_size=0.5,
    random_state=42,
    stratify=y_val_test
)

x_train = np.array(x_train)
y_train = np.array(y_train)
x_val = np.array(x_val)
y_val = np.array(y_val)
x_test = np.array(x_test)
y_test = np.array(y_test)

print("train : ",x_train.shape,y_train.shape)
print("val : ",x_val.shape,y_val.shape)
print("test : ",x_test.shape,y_test.shape)
```

```python
def prepare_data(image,label):
    image = tf.cast(image,tf.float32)
    image = preprocess_input(image)

    return image , label
```

```python
augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomRotation(0.03),
    tf.keras.layers.RandomZoom(0.10),
    tf.keras.layers.RandomTranslation(
        height_factor=0.05,
        width_factor=0.05
    )
])

def melonge_data(dataset):
  dataset = dataset.shuffle(1000 , seed=42).map(prepare_data,num_parallel_calls=tf.data.AUTOTUNE).batch(42).prefetch(tf.data.AUTOTUNE)
  return dataset

dataset_train = tf.data.Dataset.from_tensor_slices((x_train,y_train))
dataset_val = tf.data.Dataset.from_tensor_slices((x_val,y_val))
dataset_test = tf.data.Dataset.from_tensor_slices((x_test,y_test))

dataset_train = melonge_data(dataset_train)
dataset_val = melonge_data(dataset_val)
dataset_test = melonge_data(dataset_test)
```

```python
from sklearn.utils.class_weight import compute_class_weight
classes = np.unique(y_train)

weights = compute_class_weight(
    class_weight = "balanced",
    classes=classes,
    y=y_train
)

class_weights = dict(zip(classes,weights))
print(class_weights)
```

```python
base_model = tf.keras.applications.MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(224,224,3)
)

base_model.trainable = False
```

```python
def create_model(base_model):
  model = tf.keras.Sequential([
      base_model,
      tf.keras.layers.GlobalAveragePooling2D(),
      tf.keras.layers.Dense(128,activation="relu"),
      tf.keras.layers.Dropout(0.3),
      tf.keras.layers.Dense(4,activation="softmax")
  ])

  model.compile(
      optimizer="adam",
      loss="sparse_categorical_crossentropy",
      metrics=['accuracy']
  )
  
  model.summary()

  return model
```

```python
model = create_model(base_model)
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    
)

checkpoint = tf.keras.callbacks.ModelCheckpoint(
    filepath="/home/user/myfolder/model_transfer.keras",
    monitor="val_loss",
    save_best_only=False
)
```

```python
start_time = time.time()
history = model.fit(
    dataset_train,
    epochs=15,
    validation_data=dataset_val,
    callbacks = [early_stopping , checkpoint]
)
end_time= time.time()
```

```python
transfer_duration = end_time - start_time 
print("durée de Transfer Learning : ",transfer_duration)
```

```python

# Accuracy
plt.figure(figsize=(8, 5))
plt.plot(history.history["accuracy"], label="Train accuracy")
plt.plot(
    history.history["val_accuracy"],
    label="Validation accuracy"
)
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Évolution de l'accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Loss
plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], label="Train loss")
plt.plot(
    history.history["val_loss"],
    label="Validation loss"
)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Évolution de la loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
test_loss , test_accuracy = model.evaluate(dataset_test)
print("Test Loss : ",test_loss)
print("Test accuracy : ",test_accuracy)
```

## Fintening

```python
def fine_tuning(base_model,model):
    
    base_model.trainable = True
    
    for layer in base_model.layers[:-30]:
        layer.trainable = False
        
    for layer in base_model.layers[-30:]:
        layer.trainable = True   
         
    print("number de couche (True): ",sum(layer.trainable for layer in base_model.layers))
    
    model.compile(
        optimizer = tf.keras.optimizers.Adam(learning_rate=1e-5),
        loss = "sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )     
    
    return model 

```

```python
fine_tuning_model = fine_tuning(base_model,model)
```

```python
early_stoping_fine_tuning = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

checkpoint_fine_tuning = tf.keras.callbacks.ModelCheckpoint(
    monitor="val_loss",
    filepath="/home/user/myfolder/best_model.keras",
    save_best_only=True,
    mode="min"
)

backup_fine_tuning = tf.keras.callbacks.BackupAndRestore(
    backup_dir="/home/user/myfolder/backup/finetening"
)
```

```python
start_time = time.time()

fine_tuning_history = fine_tuning_model.fit(
    dataset_train,
    validation_data=dataset_val,
    epochs=15,
    callbacks=[early_stoping_fine_tuning,checkpoint_fine_tuning]
)

end_time = time.time()
```

```python

# Accuracy apre fine-tuning
plt.figure(figsize=(8, 5))
plt.plot(fine_tuning_history.history["accuracy"], label="Train accuracy")
plt.plot(
    fine_tuning_history.history["val_accuracy"],
    label="Validation accuracy"
)
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Évolution de l'accuracy")
plt.legend()
plt.grid(True)
plt.show()

# Loss fine tuning
plt.figure(figsize=(8, 5))
plt.plot(fine_tuning_history.history["loss"], label="Train loss")
plt.plot(
    fine_tuning_history.history["val_loss"],
    label="Validation loss"
)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Évolution de la loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
test_loss_fn , test_accuracy_fn = fine_tuning_model.evaluate(dataset_test)

print("Test Loss Fine-tuning : ",test_loss_fn)
print("Test Accuracy Fine-tuning : ",test_accuracy_fn)
```

```python
comparison = pd.DataFrame({
    
    "Metrics" : ["Accuracy","Loss"],
    "Avant Fine-Tuning" : [test_accuracy , test_loss],
    "Apres Fine-Tuning" : [test_accuracy_fn,test_loss_fn]
}) 

print(comparison)
```

```python

import streamlit as st
from PIL import Image, UnidentifiedImageError


MLFLOW_TRACKING_URI = os.getenv(
    "MLFLOW_TRACKING_URI",
    uri
)

MODEL_NAME = "MobileNetV2_IRM"
MODEL_ALIAS = "champion"

IMG_SIZE = (224, 224)

CLASS_NAMES = [
    "glioma",
    "meningioma",
    "notumor",
    "pituitary",
]

st.set_page_config(
    page_title="IRM Classification",
    page_icon="🧠",
    layout="centered"
)

st.title("IRM Classification")
st.write("Test du modèle champion enregistré dans MLflow.")

st.warning(
    "Prototype pédagogique uniquement. "
    "Ce résultat ne constitue pas un diagnostic médical."
)



@st.cache_resource
def load_champion():
    mlflow.set_tracking_uri(MLFLOW_TRACKING_URI)

    model_uri = f"models:/{MODEL_NAME}@{MODEL_ALIAS}"

    return mlflow.tensorflow.load_model(model_uri)



def predict_image(image, model):
    try:
        image = image.convert("RGB")
        image = image.resize(IMG_SIZE)

        array = np.asarray(image, dtype=np.float32)
        array = np.expand_dims(array, axis=0)

        predictions = np.asarray(
            model.predict(array, verbose=0)
        )

        if predictions.ndim != 2 or predictions.shape[0] != 1:
            return {"error": "Format de sortie du modèle invalide."}

        scores = predictions[0]

        if len(scores) != len(CLASS_NAMES):
            return {
                "error": (
                    "Le nombre de classes ne correspond pas "
                    "à CLASS_NAMES."
                )
            }

        if not np.all(np.isfinite(scores)):
            return {"error": "Le modèle a retourné des scores invalides."}

        index = int(np.argmax(scores))

        return {
            "predicted_class": CLASS_NAMES[index],
            "confidence": float(scores[index]),
            "scores": {
                name: float(score)
                for name, score in zip(CLASS_NAMES, scores)
            }
        }

    except Exception as error:
        return {"error": f"Échec de la prédiction : {error}"}



uploaded_file = st.file_uploader(
    "Choisir une image IRM",
    type=["jpg", "jpeg", "png", "bmp"]
)

if uploaded_file is not None:
    try:
        image = Image.open(uploaded_file)
        image.verify()

        uploaded_file.seek(0)
        image = Image.open(uploaded_file).convert("RGB")

        st.image(
            image,
            caption="Image chargée",
            width=400
        )

        if st.button("Prédire", type="primary"):
            try:
                with st.spinner("Chargement du champion depuis MLflow..."):
                    model = load_champion()

                with st.spinner("Analyse de l'image..."):
                    result = predict_image(image, model)

                if "error" in result:
                    st.error(result["error"])
                else:
                    st.subheader("Résultat")
                    st.write(
                        f"Classe prédite : **{result['predicted_class']}**"
                    )
                    st.metric(
                        "Score du modèle",
                        f"{result['confidence']:.2%}"
                    )

                    st.subheader("Scores par classe")
                    st.bar_chart(result["scores"])

            except Exception as error:
                st.error(
                    "Impossible de charger le modèle champion depuis "
                    f"MLflow. Vérifie l'accès au serveur et au Registry. "
                    f"Détail : {error}"
                )

    except (UnidentifiedImageError, OSError, ValueError):
        st.error("Le fichier est invalide ou ne peut pas être lu.")
```
